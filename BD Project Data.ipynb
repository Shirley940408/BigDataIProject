# Process NBAstats

```python
def load_nbastats(season):
    nbastats = pd.read_csv(f"{data_URL}nbastats/nbastats_{season}.csv",
                          converters={'PLAYER1_TEAM_ID': str, 'PLAYER2_TEAM_ID': str,'PLAYER3_TEAM_ID': str,
                                     'PLAYER1_ID': str, 'PLAYER2_ID': str, 'PLAYER3_ID': str})

    nbastats = nbastats[nbastats['EVENTMSGTYPE'] <= 5]
    nbastats = nbastats[nbastats['EVENTMSGTYPE'] != 4]

    nbastats['HOMEDESCRIPTION'] = nbastats['HOMEDESCRIPTION'].fillna('')
    nbastats['NEUTRALDESCRIPTION'] = nbastats['NEUTRALDESCRIPTION'].fillna('')
    nbastats['VISITORDESCRIPTION'] = nbastats['VISITORDESCRIPTION'].fillna('')
    nbastats['DESCRIPTION'] = nbastats['HOMEDESCRIPTION'].astype(str) + nbastats['NEUTRALDESCRIPTION'].astype(str) + nbastats['VISITORDESCRIPTION'].astype(str)

    nbastats.loc[nbastats['DESCRIPTION'].str.contains(' BLK\)'), 'PERSON2TYPE'] = nbastats['PERSON3TYPE']
    nbastats.loc[nbastats['DESCRIPTION'].str.contains(' BLK\)'), 'PLAYER2_ID'] = nbastats['PLAYER3_ID']
    nbastats.loc[nbastats['DESCRIPTION'].str.contains(' BLK\)'), 'PLAYER2_NAME'] = nbastats['PLAYER3_NAME']
    nbastats.loc[nbastats['DESCRIPTION'].str.contains(' BLK\)'), 'PLAYER2_TEAM_ID'] = nbastats['PLAYER3_TEAM_ID']

    nbastats = nbastats[['GAME_ID', 'EVENTNUM', 'EVENTMSGTYPE', 'PERIOD', 'PCTIMESTRING', 'DESCRIPTION', 'SCOREMARGIN',
                         'PLAYER1_ID', 'PLAYER1_NAME', 'PLAYER1_TEAM_ID', 'PLAYER1_TEAM_CITY', 'PLAYER1_TEAM_NICKNAME',
                         'PLAYER1_TEAM_ABBREVIATION', 'PLAYER2_ID', 'PLAYER2_NAME']]

    nbastats[['MINUTES', 'SECONDS']] = nbastats['PCTIMESTRING'].str.split(':', expand=True)
    nbastats[['MINUTES', 'SECONDS']] = nbastats[['MINUTES', 'SECONDS']].astype({'MINUTES': 'int', 'SECONDS': 'int'})


    nbastats['PLAYER1_TEAM_ID'] = nbastats['PLAYER1_TEAM_ID'].fillna('')
    nbastats['PLAYER2_ID'] = nbastats['PLAYER2_ID'].str.replace(r'^0$', '', regex=True)
    nbastats = nbastats[nbastats['PLAYER1_TEAM_ID'] != '']

    nbastats['SEASON'] = season

    return nbastats
```

```python
def load_shotdetail(season):
    shotdetail = pd.read_csv(f"{data_URL}shotdetail/shotdetail_{season}.csv")

    shotdetail['ACTION_TYPE'] = shotdetail['ACTION_TYPE'].str.upper()

    shotdetail['ACTION_TYPE'] = np.where(shotdetail['ACTION_TYPE'].str.contains('TURNAROUND BANK'),
                                             shotdetail['ACTION_TYPE'].str.replace('BANK', 'JUMP BANK'), shotdetail['ACTION_TYPE'])

    shotdetail['ACTION_TYPE'] = np.where(shotdetail['ACTION_TYPE'].str.contains('FADEAWAY SHOT'),
                                             shotdetail['ACTION_TYPE'].str.replace('FADEAWAY', 'JUMP FADEAWAY'), shotdetail['ACTION_TYPE'])

    # Get Any Text before Shot Method appears in String
    regex_pattern = r"^(.*?)(\b\w+ SHOT)$"
    shotdetail[['SHOT_ATTRIBUTES', 'SHOT_METHOD']] = shotdetail['ACTION_TYPE'].astype(str).str.extract(regex_pattern)

    shotdetail = shotdetail[['GAME_ID', 'GAME_EVENT_ID', 'ACTION_TYPE', 'SHOT_TYPE', 'SHOT_ZONE_BASIC', 'SHOT_ZONE_AREA', 'SHOT_ZONE_RANGE',
                               'SHOT_DISTANCE', 'LOC_X', 'LOC_Y', 'SHOT_ATTEMPTED_FLAG', 'SHOT_MADE_FLAG', 'SHOT_ATTRIBUTES', 'SHOT_METHOD', 'GAME_DATE', 'HTM', 'VTM']]

    #shotdetail['SHOT_ATTRIBUTES'] = shotdetail['SHOT_ATTRIBUTES'].str.strip()
    shotdetail['SHOT_ATTRIBUTES'] = shotdetail['SHOT_ATTRIBUTES'].str.replace("-", "")
    shotdetail["SHOT_ZONE_BASIC"] = shotdetail['SHOT_ZONE_BASIC'].str.replace("Left ", "").str.replace("Right ", "")
    shotdetail["SHOT_ZONE_AREA"]= shotdetail['SHOT_ZONE_AREA'].str.replace("Left Side Center(LC)", "Left Side")
    shotdetail["SHOT_ZONE_AREA"]= shotdetail['SHOT_ZONE_AREA'].str.replace("Right Side Center(RC)", "Right Side")
    shotdetail["SHOT_ZONE_AREA"]= shotdetail['SHOT_ZONE_AREA'].str.replace(r'\(.*?\)', '', regex=True)

    shotdetail['SHOT_ATTEMPTED_FLAG'] = shotdetail['SHOT_ATTEMPTED_FLAG'].astype(bool)
    shotdetail['SHOT_MADE_FLAG'] = shotdetail['SHOT_MADE_FLAG'].astype(bool)

    attr_cols = ["ALLEY OOP", "BANK", "CUTTING", "DRIVING", "FLOATING", "PULLUP", "PUTBACK", "RUNNING", "REVERSE", "STEP BACK", "TIP"]
    for attr in attr_cols:
        shotdetail[attr] = np.where(shotdetail['SHOT_ATTRIBUTES'].str.contains(attr), True, False)

    shotdetail = shotdetail.drop('SHOT_ATTRIBUTES', axis=1)

    return shotdetail
```

# Generate Totals for Each Season (Percentile Plots)

```python
totals = nbadata.copy()
totals = totals[totals['EVENTMSGTYPE'] <= 5]
totals['POINTS'] = np.where((totals['EVENTMSGTYPE'] == 3) & (~totals['DESCRIPTION'].str.contains(r'^MISS')), 1, 0)
totals['POINTS'] = np.where((totals['SHOT_TYPE'] == '3PT Field Goal') & (totals['SHOT_MADE_FLAG'] == 1), 3, totals['POINTS'])
totals['POINTS'] = np.where((totals['SHOT_TYPE'] == '2PT Field Goal') & (totals['SHOT_MADE_FLAG'] == 1), 2, totals['POINTS'])

points = totals.groupby(['SEASON', 'PLAYER1_ID'])['POINTS'].sum()
FTA = totals[totals['EVENTMSGTYPE'] == 3].groupby(['SEASON', 'PLAYER1_ID'])['EVENTMSGTYPE'].count()
FGA = totals[totals['EVENTMSGTYPE'] < 3].groupby(['SEASON', 'PLAYER1_ID'])['SHOT_DISTANCE'].agg(['mean', 'count'])
AST = totals[(totals['EVENTMSGTYPE'] < 3) & (totals['SHOT_MADE_FLAG'] == 1) & (totals['PLAYER2_ID'] != 0)].groupby(['SEASON', 'PLAYER2_ID'])['EVENTMSGTYPE'].count()
TOV = totals[totals['EVENTMSGTYPE'] == 5].groupby(['SEASON', 'PLAYER1_ID'])['EVENTMSGTYPE'].count()

points = points.reset_index().rename(columns={'PLAYER1_ID': 'ID'})
FTA = FTA.reset_index().rename(columns={'PLAYER1_ID': 'ID', 'EVENTMSGTYPE': 'FTA'})
FGA = FGA.reset_index().rename(columns={'PLAYER1_ID': 'ID', 'mean': 'SHOT_DISTANCE', 'count' : 'FGA'})
AST = AST.reset_index().rename(columns={'PLAYER2_ID': 'ID', 'EVENTMSGTYPE': 'AST'})
TOV = TOV.reset_index().rename(columns={'PLAYER1_ID': 'ID', 'EVENTMSGTYPE': 'TOV'})

totals = pd.merge(points, FTA, on=['SEASON', 'ID'], how='outer')
totals = pd.merge(totals, FGA, on=['SEASON', 'ID'], how='outer')
totals = pd.merge(totals, AST, on=['SEASON', 'ID'], how='outer')
totals = pd.merge(totals, TOV, on=['SEASON', 'ID'], how='outer')

totals = totals.fillna(0)
totals['PPP'] = totals['POINTS']/(totals['FGA']+0.44*totals['FTA']+totals['TOV'])
totals['TS%'] = totals['POINTS']/(2*(totals['FGA']+0.44*totals['FTA']))*100
totals['AST/TOV'] = totals['AST']/totals['TOV']
totals['FTr'] = totals['FTA']/totals['FGA']
totals
```

```python
totals.to_parquet(
    f"{data_URL}sample/season_totals",
    engine='pyarrow',
    partition_cols=['SEASON'],
    index=False
)
```
